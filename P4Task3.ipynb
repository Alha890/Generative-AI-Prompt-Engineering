```python
!pip install groq

from groq import Groq

client = Groq(
    api_key="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"
)

topic = "Artificial Intelligence"

outline_prompt = f"""
Generate a blog outline with 5 headings for the topic:
{topic}
"""

outline_response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "user",
            "content": outline_prompt
        }
    ],
    temperature=0.5,
    max_tokens=300
)

outline = outline_response.choices[0].message.content

print("=" * 70)
print("BLOG OUTLINE")
print("=" * 70)
print(outline)

content_prompt = f"""
Write 2 detailed paragraphs for each heading below.

Headings:
{outline}
"""

content_response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "user",
            "content": content_prompt
        }
    ],
    temperature=0.6,
    max_tokens=1200
)

content = content_response.choices[0].message.content

print("\n" + "=" * 70)
print("BLOG CONTENT")
print("=" * 70)
print(content)

intro_conclusion_prompt = f"""
Write:
1. A 3-sentence introduction
2. A 3-sentence conclusion

for a blog post on:
{topic}
"""

intro_conclusion_response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {
            "role": "user",
            "content": intro_conclusion_prompt
        }
    ],
    temperature=0.5,
    max_tokens=300
)

intro_conclusion = intro_conclusion_response.choices[0].message.content

print("\n" + "=" * 70)
print("INTRODUCTION & CONCLUSION")
print("=" * 70)
print(intro_conclusion)

full_blog = f"""
BLOG TITLE: {topic}

{intro_conclusion}

{content}
"""

with open("blog_post.txt", "w", encoding="utf-8") as file:
    file.write(full_blog)

print("\n" + "=" * 70)
print("BLOG POST SAVED SUCCESSFULLY")
print("=" * 70)
print("File Name: blog_post.txt")
```
