```python

facts = {
    "population of india": "1.44 billion",
    "speed of light": "3x10^8 m/s"
}


def lookup(query):

    query = query.lower()

    if query in facts:
        return facts[query]

    return "Fact not found"



def calculate(expression):

    try:
        return eval(expression)

    except Exception as e:
        return f"Error: {e}"


def react_agent(problem):

    print("=" * 60)
    print("PROBLEM")
    print("=" * 60)
    print(problem)

    print("\nThought: I need to lookup the population of India first.")

    observation1 = lookup("population of india")

    print("Action: lookup('population of india')")
    print("Observation:", observation1)

    print("\nThought: Now I will calculate 10% of the population.")

    # 1.44 billion = 1.44 * 1000000000
    population = 1.44 * 1000000000

    observation2 = calculate(f"{population} * 0.10")

    print("Action: calculate(population * 0.10)")
    print("Observation:", observation2)

    print("\nFinal Answer:")
    print(f"10% of India's population is approximately {observation2}")



problem = """
Find the population of India and calculate 10% of it.
"""

react_agent(problem)
```
